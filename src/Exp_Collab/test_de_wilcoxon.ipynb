```r
# ======================
# Librerías necesarias
# ======================
library(readxl)
library(dplyr)
library(ggplot2)
library(ggpubr)

# ======================
# 1. Leer los datos
# ======================
data <- read_excel("data.xlsx")

# Asegurarse de que la variable sea categórica
data$`Tipo de Data Drifting` <- as.factor(data$`Tipo de Data Drifting`)

# ======================
# 2. Test de Wilcoxon por pares
# ======================
wilcox_results <- pairwise.wilcox.test(
  x = data$`Mejor restultado`,
  g = data$`Tipo de Data Drifting`,
  p.adjust.method = "holm"
)

print(wilcox_results)

# ======================
# 3. Boxplots con p-values
# ======================
# Sacar todas las combinaciones posibles de comparaciones
comparisons <- combn(levels(data$`Tipo de Data Drifting`), 2, simplify = FALSE)

# Graficar
ggplot(data, aes(x = `Tipo de Data Drifting`, y = `Mejor restultado`, fill = `Tipo de Data Drifting`)) +
  geom_boxplot(alpha = 0.7, outlier.color = "red", outlier.shape = 16) +
  geom_jitter(width = 0.2, alpha = 0.5) +
  labs(title = "Comparación de Mejor resultado por Tipo de Data Drifting",
       x = "Tipo de Data Drifting",
       y = "Mejor resultado") +
  theme_minimal() +
  theme(legend.position = "none") +
  stat_compare_means(comparisons = comparisons, method = "wilcox.test", 
                     label = "p.signif", hide.ns = TRUE)  # agrega p-values significativos
```
